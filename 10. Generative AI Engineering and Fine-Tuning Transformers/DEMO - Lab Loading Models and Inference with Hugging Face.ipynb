```python
from transformers import pipeline
from transformers import DistilBertForSequenceClassification, DistilBertTokenizer
import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer

def warn(*args, **kwargs):
    pass

import warnings
warnings.warn = warn
warnings.filterwarnings('ignore')
```

```python
tokenizer = DistilBertTokenizer.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
model = DistilBertForSequenceClassification.from_pretrained("distilbert-base-uncased-finetuned-sst-2-english")
```

```python
text =  "Congratulations! You've won a free ticket to the Bahamas. Reply WIN to claim."

inputs = tokenizer(text, return_tensors='pt')
print(inputs)
```

```python
with torch.no_grad():
    outputs = model(**inputs)
```

```python
print(outputs)
```

```python
logits = outputs.logits
print(logits)
```

```python
logits.shape
```

```python
probabilities = torch.softmax(logits, dim=-1)
probabilities
```

```python
predicted_class = torch.argmax(probabilities, dim=-1)
print(predicted_class)
```

```python
labels = ["NEGATIVE", "POSITIVE"]
predicted_label = labels[predicted_class]
predicted_label
```

# Text generation with GPT-2


```python
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")
```

```python
prompt = "once upon a time"
inputs = tokenizer(prompt, return_tensors='pt')
print(inputs)
```

```python
output_ids = model.generate(
    inputs.input_ids,
    attention_mask=inputs.attention_mask,
    pad_token_id=tokenizer.eos_token_id,
    max_length=50,
    num_return_sequences=1
)
output_ids
```

```python
with torch.no_grad():
    outputs = model(**inputs)
outputs
```

```python
generated_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)
generated_text
```

```python
classifier = pipeline(
    'text-classification',
    model='distilbert-base-uncased-finetuned-sst-2-english'
)
result = classifier("Congratulations! You've won a free ticket to the Bahamas. Reply WIN to claim.")
result
```

```python
classifier = pipeline("text-classification",
                      model="papluca/xlm-roberta-base-language-detection")
result = classifier("Bonjour, comment ça va?")
result
```

```python
generator = pipeline(
    'text-generation',
    model='gpt2'
)

prompt = 'Once upon a time'
result = generator(
    prompt,
    max_length=50,
    num_return_sequences=1,
    truncation=True
)
```

```python
generator = pipeline(
    'text2text-generation',
    model='t5-small'
)
prompt = 'translate English to French: How are you?'
result = generator(
    prompt,
    max_length=50,
    num_return_sequences=1
)
result
```
