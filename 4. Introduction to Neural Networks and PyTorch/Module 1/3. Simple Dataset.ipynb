```python
import torch
from torch.utils.data import Dataset
torch.manual_seed(1)
```

```python
class toy_set(Dataset):
    def __init__(self, length=100, transform= None):
        self.len = length
        self.x = 2 * torch.ones(length, 2)
        self.y = torch.ones(length, 1)
        self.transform = transform

    def __getitem__(self, index):
        sample = self.x[index], self.y[index]
        if self.transform:
            sample = self.transform(sample)
        return sample

    def __len__(self):
        return self.len
```

```python
our_dataset = toy_set()

print(our_dataset)
print(our_dataset[0])
print(len(our_dataset))
```

```python
for i in range(3):
    x, y = our_dataset[i]
    print(f"Index: {i}, x: {x}, y: {y}")
```

```python
class addition_multiplication(object):

    def __init__(self, addition_x = 1, multiplication_y = 2):
        self.addition_x = addition_x
        self.multiplication_y = multiplication_y

    def __call__(self, sample):
        x = sample[0]
        y = sample[1]
        x = x + self.addition_x
        y = y * self.multiplication_y
        sample = x, y
        return sample
```

```python
addition_multiplication_object = addition_multiplication()
dataset = toy_set()

for i in range(10):
    x, y = dataset[i]
    print(f"Original values: x >> {x}, y >> {y}")
    x_, y_ = addition_multiplication_object(dataset[i])
    print(f"Transformed values: x_ >> {x_}, y >> {y_}")
```

```python
dataset_with_transform = toy_set(transform=addition_multiplication_object)
for i in range(10):
    x, y = dataset[i]
    print(f"Original values: x >> {x}, y >> {y}")
    x_, y_ = dataset_with_transform[1]
    print(f"Transformed values: x_ >> {x_}, y_ >> {y_}")
```

```python
from torchvision import transforms
class multiplication(object):
    def __init__(self, multiply = 100):
        self.multiply = multiply
    def __call__(self, sample):
        x = sample[0]
        y = sample[1]
        x = x * self.multiply
        y = y * self.multiply
        sample = x, y
        return sample
```

```python
data_transform = transforms.Compose([addition_multiplication(), multiplication()])
print(f"Combination of transforms: {data_transform}")
```

```python
data_transform(dataset[0])
```

```python
x, y = dataset[0]
x_, y_ = data_transform(dataset[0])

print(f"Original values: {x} - {y}")
print(f"Transformed values: {x_} - {y_}")
```

```python
compose_dataset = toy_set(transform=data_transform)

for i in range(3):
    x, y = dataset[i]
    x_, y_ = dataset_with_transform[i]
    x_co, y_co = compose_dataset[i]

    print(f"Originals: {x} - {y}")
    print(f"Transformed: {x_}, {y_}")
    print(f"Compose: {x_co} - {y_co}")
```
