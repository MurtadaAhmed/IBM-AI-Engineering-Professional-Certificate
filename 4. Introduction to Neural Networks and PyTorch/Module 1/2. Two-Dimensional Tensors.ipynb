```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import pandas as pd
```

```python
twoD_list = [[11, 12, 13], [21, 22, 23], [31, 32, 33]]
twoD_tensor = torch.tensor(twoD_list)
print(twoD_tensor)
```

```python
print(twoD_tensor.ndimension())
print(twoD_tensor.size())
print(twoD_tensor.shape)
print(twoD_tensor.numel())
```

```python
twoD_numpy = twoD_tensor.numpy()
print(twoD_numpy)
print(twoD_numpy.dtype)
```

```python
new_twoD_tensor = torch.from_numpy(twoD_numpy)
print(new_twoD_tensor)
print(new_twoD_tensor.dtype)
```

```python
df = pd.DataFrame({'a':[11,21,31],'b':[12,22,312]})
print(df.values)
print(df.values.dtype)
```

```python
new_tensor = torch.from_numpy(df.values)
print(new_tensor)
print(new_tensor.dtype)
```

```python
df = pd.DataFrame({'A':[11, 33, 22],'B':[3, 3, 2]})
new_tensor = torch.from_numpy(df.values)
print(new_tensor)
```

```python
Y = torch.tensor([[2, 1], [1, 2]])
two_Y = 2 * Y
print(two_Y)
```

```python
A = torch.tensor([[0, 1, 1], [1, 0, 1]])
B = torch.tensor([[1, 1], [1, 1], [-1, 1]])
A_times_B = torch.mm(A, B)
print(A_times_B)
```
