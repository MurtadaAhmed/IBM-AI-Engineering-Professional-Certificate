```python
import torch
import matplotlib.pylab as plt
import numpy as np
torch.manual_seed(0)
```

```python
def show_data(data_sample, shape = (28, 28)):
    plt.imshow(data_sample[0].numpy().reshape(shape), cmap='gray')
    plt.title('y = ' + str(data_sample[1]))
```

```python
import torchvision.transforms as transforms
import torchvision.datasets as dsets
```

```python
dataset = dsets.MNIST(
    root = '',
    download = True,
    transform = transforms.ToTensor()
)
```

```python
print(type(dataset[0]))
print(len(dataset[0]))
print(dataset[0][0].shape)
print(type(dataset[0][0]))
```

```python
croptensor_data_transform = transforms.Compose([transforms.CenterCrop(20), transforms.ToTensor()])
dataset = dsets.MNIST(root = '', download = True, transform = croptensor_data_transform)
dataset[0][0].shape
```

```python
show_data(dataset[0],shape = (20, 20))
```

```python
show_data(dataset[1],shape = (20, 20))
```

```python
fliptensor_data_transform = transforms.Compose([transforms.RandomHorizontalFlip(p = 1),transforms.ToTensor()])
dataset = dsets.MNIST(root = '', download = True, transform = fliptensor_data_transform)
show_data(dataset[1])
```

```python
my_data_transform = transforms.Compose([transforms.RandomVerticalFlip(p = 1), transforms.RandomHorizontalFlip(p = 1), transforms.ToTensor()])
dataset = dsets.MNIST(root = '', train = False, download = True, transform = my_data_transform)
show_data(dataset[1])
```
